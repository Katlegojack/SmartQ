## My embedded dataset details

I embedded the following dataset inside this notebook:

- **Rows:** 100,000
- **Columns:** 45
- **CSV name:** `SmartQ_Synthetic_Operational_Dataset_100k.csv`
- **Original CSV size:** 33.53 MB
- **Compressed payload size:** 4.37 MB
- **SHA-256:** `616a25c43475fa44c0e8c17521a1156603aab99aea22802a319542b50b374b3e`

I use the SHA-256 value as an integrity check so I can confirm that the extracted CSV matches the dataset I originally embedded.


```python
from pathlib import Path
import base64
import gzip
import hashlib

DATASET_FILENAME = "SmartQ_Synthetic_Operational_Dataset_100k.csv"
EXPECTED_SHA256 = "616a25c43475fa44c0e8c17521a1156603aab99aea22802a319542b50b374b3e"

dataset_path = Path(DATASET_FILENAME)

if not dataset_path.exists():
    compressed_bytes = base64.b64decode(SMARTQ_DATASET_GZIP_BASE64)
    csv_bytes = gzip.decompress(compressed_bytes)
    dataset_path.write_bytes(csv_bytes)
    print(f"Dataset extracted to: {dataset_path.resolve()}")
else:
    print(f"Dataset already exists: {dataset_path.resolve()}")

actual_sha256 = hashlib.sha256(dataset_path.read_bytes()).hexdigest()
assert actual_sha256 == EXPECTED_SHA256, "Dataset checksum mismatch."
print("Dataset integrity check passed.")
```

```python
import pandas as pd

df = pd.read_csv(dataset_path)

print(f"Rows: {len(df):,}")
print(f"Columns: {df.shape[1]}")
display(df.head())
```

```python
print("Dataset shape:", df.shape)

print("\nStatus counts:")
display(df["status"].value_counts(dropna=False))

print("\nQueue type counts:")
display(df["queue_type"].value_counts(dropna=False))

print("\nService counts:")
display(df["service_name"].value_counts(dropna=False))

print("\nMissing values in key modelling fields:")
key_fields = [
    "actual_wait_minutes",
    "people_ahead",
    "effective_open_counters",
    "queue_pressure_index",
    "recent_avg_service_minutes_10",
    "service_target_minutes",
    "hour_of_day",
    "day_of_week",
]
display(df[key_fields].isna().sum().to_frame("missing_rows"))
```

## My most important modelling rule

My prediction target is:

`actual_wait_minutes`

For waiting-time regression, I use **completed visits only**.

I exclude fields such as `call_time`, `actual_wait_minutes`, `wait_variance_minutes`, `actual_service_minutes`, `service_started_at`, and `service_completed_at` from model inputs.

I do this because those fields contain information from after the prediction point or directly contain the answer.

If I used them, I would create **data leakage**, which would make the model look unrealistically accurate.
